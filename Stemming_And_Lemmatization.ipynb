```python
## Classification Problem
## Comments of product is a positive review or negative review
## Reviews----> eating, eat,eaten [going,gone,goes]--->go

words=["eating","eats","eaten","writing","writes","programming","programs","history","finally","finalized"]
```

```python
from nltk.stem import PorterStemmer
stemmer = PorterStemmer()
```

```python
for word in words:
    print(word,"====>",stemmer.stem(word))
```

```python
stemmer.stem('congratulations')
```

```python
stemmer.stem('sitting')
```

```python
from nltk.stem import RegexpStemmer
regstemmer = RegexpStemmer('ing$|e$|s$|able$', min=4) # removes ing, s, e, able from the last, it will not consider the starting and middle position
```

```python
regstemmer.stem('eating')
```

```python
from nltk.stem import SnowballStemmer
snowballstemmer = SnowballStemmer('english')
```

```python
for word in words:
    print(word+"---->",snowballstemmer.stem(word))
```

```python
stemmer.stem('sportingly'), stemmer.stem('fairly')
```

```python
snowballstemmer.stem('sportingly'), snowballstemmer.stem('fairly')
```

```python
snowballstemmer.stem('goes'), stemmer.stem('goes')
```

```python
from nltk.stem import WordNetLemmatizer
lematizer = WordNetLemmatizer()
lematizer.lemmatize('going',pos='v')
```

```python
for word in words:
    print(word+"--->",lematizer.lemmatize(word, pos='v'))
```

```python
lematizer.lemmatize('sportingly', pos='v'), lematizer.lemmatize('fairly', pos='v')
```
